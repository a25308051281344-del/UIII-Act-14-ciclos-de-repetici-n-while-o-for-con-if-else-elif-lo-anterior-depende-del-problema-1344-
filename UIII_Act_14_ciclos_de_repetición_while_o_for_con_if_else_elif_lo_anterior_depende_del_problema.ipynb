#### 29.Programa que muestre un triángulo invertido de asteriscos de altura N.

```python
print("=== TRIÁNGULO INVERTIDO DE ASTERISCOS ===")

# Pedir la altura
n = int(input("Ingresa la altura del triángulo: "))

# Variable de control
fila = n

# Ciclo while
while fila >= 1:

    # Variable para imprimir asteriscos
    columna = 1

    # Imprimir asteriscos en la fila
    while columna <= fila:
        print("*", end=" ")
        columna += 1

    # Salto de línea
    print()

    # Disminuir fila
    fila -= 1
```
